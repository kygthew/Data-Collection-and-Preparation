```python
df["order_date_parsed"] = pd.to_datetime(
    df["order_date"],
    errors="coerce"
)

print("Invalid dates:",
      df["order_date_parsed"].isna().sum())
```

```python
print(df["quantity"].describe())
```

```python
print("Negative quantity:",
      (df["quantity"] < 0).sum())
```

```python
print("Invalid ages:",
      ((df["customer_age"] < 0) |
       (df["customer_age"] > 120)).sum())
```

```python
print("Invalid ratings:",
      ((df["customer_rating"] < 1) |
       (df["customer_rating"] > 5)).sum())
```

```python
Q1 = df["net_sales"].quantile(0.25)
Q3 = df["net_sales"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["net_sales"] < lower) |
    (df["net_sales"] > upper)
]

print("Number of outliers:", len(outliers))
```

```python
df_clean = df.copy()
```

```python
df_clean["order_date"] = pd.to_datetime(
    df_clean["order_date"],
    errors="coerce"
)
```

```python
df_clean["gender"] = df_clean["gender"].str.strip().str.lower()
```

```python
df_clean.info()
```

```python
df_clean.to_csv(
    "ecommerce_clean.csv",
    index=False
)
```

```python
df_clean.to_parquet(
    "ecommerce_clean.parquet",
    index=False
)
```

```python
start = time.perf_counter()

csv_test = pd.read_csv("test.csv")

csv_read_time = time.perf_counter() - start

#Parquet
start = time.perf_counter()

parquet_test = pd.read_parquet("test.parquet")

parquet_read_time = time.perf_counter() - start
```
