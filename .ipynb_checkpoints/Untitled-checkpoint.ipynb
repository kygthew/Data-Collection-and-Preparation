```python
    # Task 1–5: Data Transformation & EDA
# Dataset: E-Commerce Sales Analytics Dataset — 150K+ Transactions | 2021–2025
# Source: https://huggingface.co/datasets/shera21/e-commerce-sales-and-customer-analytics

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 30)

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("ecommerce_clean.csv")

print("Dataset loaded successfully!")
print("Dataset name: E-Commerce Sales Analytics Dataset")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df.head())
```

```python
# Initial inspection

print("Dataset information:")
df.info()

print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
display(
    df.isnull().sum()
      .sort_values(ascending=False)
      .head(15)
      .to_frame("Missing values")
)

print("\nDescriptive statistics:")
display(df.describe(include="all").T)
```

```python
# Select numerical columns for aggregation

numeric_columns = [
    "net_sales",
    "profit",
    "gross_sales",
    "quantity",
    "customer_rating",
    "delivery_days"
]

available_numeric = [
    col for col in numeric_columns if col in df.columns
]

# Convert selected columns to numeric values
for col in available_numeric:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Group by one categorical column
grouped_segment = df.groupby(
    "customer_segment",
    dropna=False
).agg(
    transaction_count=("order_id", "count"),
    unique_customers=("customer_id", "nunique"),
    average_net_sales=("net_sales", "mean"),
    total_net_sales=("net_sales", "sum"),
    average_profit=("profit", "mean"),
    total_profit=("profit", "sum"),
    average_rating=("customer_rating", "mean")
).round(2)

print("Task 1: GroupBy customer_segment")
display(grouped_segment)
```

```python
# Group by two categorical columns
# Apply multiple aggregation functions

grouped_region_segment = df.groupby(
    ["region", "customer_segment"],
    dropna=False
).agg(
    transaction_count=("order_id", "count"),
    total_sales=("net_sales", "sum"),
    average_sales=("net_sales", "mean"),
    total_profit=("profit", "sum"),
    average_profit=("profit", "mean"),
    average_delivery_days=("delivery_days", "mean")
).round(2)

print("Summary by region and customer segment:")
display(grouped_region_segment)
```

```python
# Create an orders DataFrame
orders_df = df[
    [
        "order_id",
        "customer_id",
        "net_sales",
        "profit",
        "order_status"
    ]
].copy()

# Create a customer DataFrame
customer_columns = [
    "customer_id",
    "customer_name",
    "customer_city",
    "customer_country"
]

customer_columns = [
    col for col in customer_columns if col in df.columns
]

customers_df = (
    df[customer_columns]
    .dropna(subset=["customer_id"])
    .drop_duplicates(subset=["customer_id"])
    .copy()
)

print("Orders DataFrame:")
display(orders_df.head())

print("Customers DataFrame:")
display(customers_df.head())

print("Orders rows:", len(orders_df))
print("Unique customers:", customers_df["customer_id"].nunique())
```

```python
# INNER JOIN
inner_join = orders_df.merge(
    customers_df,
    on="customer_id",
    how="inner",
    indicator=True
)

# LEFT JOIN
left_join = orders_df.merge(
    customers_df,
    on="customer_id",
    how="left",
    indicator=True
)

print("INNER JOIN result:")
display(inner_join.head(10))

print("LEFT JOIN result:")
display(left_join.head(10))

# Compare results
join_comparison = pd.DataFrame({
    "Join type": ["Inner join", "Left join"],
    "Number of rows": [len(inner_join), len(left_join)],
    "Matched rows": [
        (inner_join["_merge"] == "both").sum(),
        (left_join["_merge"] == "both").sum()
    ],
    "Unmatched rows": [
        (inner_join["_merge"] != "both").sum(),
        (left_join["_merge"] != "both").sum()
    ]
})

print("Comparison of joins:")
display(join_comparison)

print(
    "Left join preserves all rows from orders_df, "
    "while inner join keeps only rows with matching customer IDs."
)
```

```python
# Show rows that did not find a matching customer
unmatched_orders = left_join[
    left_join["_merge"] == "left_only"
]

print("Number of unmatched orders:", len(unmatched_orders))
display(unmatched_orders.head())

# Remove the merge indicator from the final joined DataFrames
inner_join = inner_join.drop(columns=["_merge"])
left_join = left_join.drop(columns=["_merge"])
```

```python
# Pivot table: average net sales by region and sales channel

pivot_data = df.pivot_table(
    index="region",
    columns="sales_channel",
    values="net_sales",
    aggfunc="mean",
    fill_value=0
).round(2)

print("Pivot table: Average net sales by region and sales channel")
display(pivot_data)
```

```python
# Melt: transform wide data into long format

melt_columns = [
    "customer_segment",
    "net_sales",
    "profit"
]

melt_data = df[melt_columns].head(10).copy()

melted_df = pd.melt(
    melt_data,
    id_vars=["customer_segment"],
    value_vars=["net_sales", "profit"],
    var_name="metric",
    value_name="value"
)

print("Original wide-format DataFrame:")
display(melt_data)

print("Transformed long-format DataFrame:")
display(melted_df)
```

```python
# Plot 1: Distribution of a numerical variable
# Histogram of net_sales

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="net_sales",
    bins=40,
    kde=True
)

plt.title("Distribution of Net Sales")
plt.xlabel("Net Sales")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# Plot 2: Comparison of categories
# Average net sales by customer segment

segment_sales = (
    df.groupby("customer_segment", dropna=False)["net_sales"]
      .mean()
      .sort_values(ascending=False)
      .reset_index()
)

plt.figure(figsize=(10, 5))

sns.barplot(
    data=segment_sales,
    x="customer_segment",
    y="net_sales"
)

plt.title("Average Net Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Average Net Sales")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Plot 3: Relationship between two numerical variables
# Gross sales vs net sales

scatter_data = df[
    ["gross_sales", "net_sales"]
].dropna()

# Sample data for a clearer plot when the dataset is large
if len(scatter_data) > 5000:
    scatter_data = scatter_data.sample(
        n=5000,
        random_state=42
    )

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=scatter_data,
    x="gross_sales",
    y="net_sales",
    alpha=0.5
)

plt.title("Relationship Between Gross Sales and Net Sales")
plt.xlabel("Gross Sales")
plt.ylabel("Net Sales")
plt.tight_layout()
plt.show()
```

```python
# Plot 4: Compare a numerical variable across categories
# Profit distribution by region

plt.figure(figsize=(11, 6))

sns.boxplot(
    data=df,
    x="region",
    y="profit"
)

plt.title("Profit Distribution Across Regions")
plt.xlabel("Region")
plt.ylabel("Profit")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# EDA 1: Customer segment with the highest average net sales

segment_stats = (
    df.groupby("customer_segment", dropna=False)
      .agg(
          average_net_sales=("net_sales", "mean"),
          total_net_sales=("net_sales", "sum"),
          average_profit=("profit", "mean"),
          transaction_count=("order_id", "count")
      )
      .round(2)
)

best_segment = segment_stats["average_net_sales"].idxmax()

print("Finding 1: Customer segments")
print(
    f"The segment with the highest average net sales is "
    f"{best_segment}, with average net sales of "
    f"{segment_stats.loc[best_segment, 'average_net_sales']:.2f}."
)

display(segment_stats.sort_values(
    "average_net_sales",
    ascending=False
))
```

```python
# EDA 2: Region with the highest total net sales

region_stats = (
    df.groupby("region", dropna=False)
      .agg(
          total_net_sales=("net_sales", "sum"),
          average_net_sales=("net_sales", "mean"),
          total_profit=("profit", "sum"),
          average_delivery_days=("delivery_days", "mean")
      )
      .round(2)
)

best_region = region_stats["total_net_sales"].idxmax()

print("Finding 2: Regional performance")
print(
    f"The region with the highest total net sales is "
    f"{best_region}, with total net sales of "
    f"{region_stats.loc[best_region, 'total_net_sales']:.2f}."
)

display(region_stats.sort_values(
    "total_net_sales",
    ascending=False
))
```

```python
# EDA 3: Relationship between gross sales and net sales

sales_correlation = df[
    ["gross_sales", "net_sales"]
].corr().iloc[0, 1]

print("Finding 3: Sales relationship")
print(
    f"The Pearson correlation between gross sales and net sales "
    f"is {sales_correlation:.3f}."
)

if pd.isna(sales_correlation):
    print("The correlation could not be calculated from the available data.")
elif sales_correlation >= 0.7:
    print("This indicates a strong positive linear relationship.")
elif sales_correlation >= 0.3:
    print("This indicates a moderate positive linear relationship.")
elif sales_correlation > 0:
    print("This indicates a weak positive linear relationship.")
elif sales_correlation <= -0.7:
    print("This indicates a strong negative linear relationship.")
elif sales_correlation <= -0.3:
    print("This indicates a moderate negative linear relationship.")
else:
    print("The linear relationship is weak or close to zero.")
```
