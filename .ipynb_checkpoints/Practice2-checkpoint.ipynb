```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", 50)
pd.set_option("display.max_rows", 20)
pd.set_option("display.width", 120)
pd.set_option("display.float_format", lambda x: f"{x:.2f}")

print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)
```

```python
# Create a two-dimensional NumPy array
arr = np.array([
    [120,  80,  30,  5],
    [150,  90,  45,  7],
    [100,  70,  25,  4],
    [200, 110,  60,  9],
    [175,  95,  50,  6]
])

print("NumPy array:")
print(arr)
```

```python
print("Shape:", arr.shape)
print("Number of dimensions:", arr.ndim)
print("Total number of elements:", arr.size)
print("Data type:", arr.dtype)
print("Number of rows:", arr.shape[0])
print("Number of columns:", arr.shape[1])
```

```python
# Select individual elements
print("First element:", arr[0, 0])
print("Element in row 2, column 3:", arr[1, 2])
print("Last element:", arr[-1, -1])

# Select rows
print("\nFirst row:")
print(arr[0])

print("\nThird row:")
print(arr[2])

# Select columns
print("\nFirst column:")
print(arr[:, 0])

print("\nThird column:")
print(arr[:, 2])
```

```python
# First three rows
print("First three rows:")
print(arr[:3, :])

# First two columns
print("\nFirst two columns:")
print(arr[:, :2])

# A smaller subarray: rows 2-4 and columns 2-3
print("\nSmaller subarray:")
print(arr[1:4, 1:3])

# Every second row
print("\nEvery second row:")
print(arr[::2, :])
```

```python
print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))
print("Standard deviation:", np.std(arr))
```

```python
print("Column sums (axis=0):")
print(np.sum(arr, axis=0))

print("\nRow sums (axis=1):")
print(np.sum(arr, axis=1))

print("\nColumn means (axis=0):")
print(np.mean(arr, axis=0))

print("\nRow means (axis=1):")
print(np.mean(arr, axis=1))
```

```python
print("Original array:")
print(arr)

print("\naxis=0: operation down the rows, producing one result per column")
print(np.sum(arr, axis=0))

print("\naxis=1: operation across the columns, producing one result per row")
print(np.sum(arr, axis=1))
```

```python
# Add a scalar to every element
print("Array + 10:")
print(arr + 10)

# Multiply every element by 2
print("\nArray * 2:")
print(arr * 2)

# Element-wise addition
print("\nArray + Array:")
print(arr + arr)

# Element-wise multiplication
print("\nArray * Array:")
print(arr * arr)

# Element-wise comparison
print("\nElements greater than 100:")
print(arr > 100)

# Conditional selection
print("\nValues greater than 100:")
print(arr[arr > 100])

# Normalize each column using its mean and standard deviation
column_means = arr.mean(axis=0)
column_stds = arr.std(axis=0)

arr_standardized = (arr - column_means) / np.where(column_stds == 0, 1, column_stds)

print("\nStandardized array:")
print(np.round(arr_standardized, 2))
```

```python
# Load the dataset
df = pd.read_csv("ecommerce_clean.csv")

print("Dataset loaded successfully!")
print("First 5 rows:")
display(df.head())
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("Dataset shape:", df.shape)
print("Total number of cells:", df.size)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataFrame information:")
df.info()
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns.tolist()

categorical_columns = df.select_dtypes(
    include=["object", "category", "bool", "string"]
).columns.tolist()

datetime_columns = df.select_dtypes(include=["datetime", "datetimetz"]).columns.tolist()

print("Numerical variables:")
print(numeric_columns)

print("\nCategorical variables:")
print(categorical_columns)

print("\nDatetime variables:")
print(datetime_columns)

print("\nNumber of numerical variables:", len(numeric_columns))
print("Number of categorical variables:", len(categorical_columns))
print("Number of datetime variables:", len(datetime_columns))
```

```python
missing_counts = df.isnull().sum()

missing_summary = pd.DataFrame({
    "Missing values": missing_counts,
    "Missing percentage (%)": (missing_counts / len(df) * 100).round(2)
})

display(missing_summary)

print("Total missing values:", int(df.isnull().sum().sum()))

if df.isnull().sum().sum() == 0:
    print("There are no missing values in the dataset.")
else:
    print("The dataset contains missing values.")
```

```python
print("Descriptive statistics for numerical variables:")
display(df.describe())

print("\nDescriptive statistics for all variables:")
display(df.describe(include="all").T)
```

```python
print("First 5 records:")
display(df.head())

print("\nLast 5 records:")
display(df.tail())

print("\nRandom sample of 5 records:")
display(df.sample(n=min(5, len(df)), random_state=42))
```

```python
print("Dataset columns:")
print(df.columns.tolist())

if "order_id" in df.columns:
    print("\nOne row represents an e-commerce transaction or order record.")
    print("The order_id column identifies the order.")
else:
    print("\nOne row represents one record in the E-Commerce Sales Analytics Dataset.")

print("\nExample record:")
display(df.head(1))
```

```python
# Select one column as a Series
if "net_sales" in df.columns:
    print("Net sales:")
    display(df["net_sales"].head())

# Select multiple columns as a DataFrame
preferred_columns = [
    "order_id",
    "order_date",
    "region",
    "gender",
    "net_sales",
    "profit"
]

selected_columns = [col for col in preferred_columns if col in df.columns]

print("\nSelected columns:")
display(df[selected_columns].head())
```

```python
# Select the first row by integer position
print("First row:")
display(df.iloc[0])

# Select the first five rows
print("\nFirst five rows:")
display(df.iloc[:5])

# Select the first three rows and first four columns
print("\nFirst three rows and first four columns:")
display(df.iloc[:3, :4])

# Select the first row and first column as a single value
print("\nValue at row position 0, column position 0:")
print(df.iloc[0, 0])

# Select the last three rows
print("\nLast three rows:")
display(df.iloc[-3:])
```

```python
# Select rows using index labels
print("Rows with index labels 0, 1, and 2:")
display(df.loc[0:2])

# Select specific columns using their names
print("\nSelected columns:")
display(df.loc[0:4, selected_columns])

# Select a single value by index label and column name
first_column = df.columns[0]
print("\nSingle value:")
print(df.loc[0, first_column])

# Select rows using a condition
if "region" in df.columns:
    example_region = df["region"].dropna().iloc[0] if df["region"].notna().any() else None

    if example_region is not None:
        print(f"\nRows where region = {example_region}:")
        display(df.loc[df["region"] == example_region].head(10))
```

```python
print("Using loc: index labels 0 through 4, inclusive")
display(df.loc[0:4, selected_columns])

print("\nUsing iloc: row positions 0 through 4, excluding position 5")
display(df.iloc[0:5, :][selected_columns])
```

```python
# Question 1:
# Which orders have net_sales greater than the average net_sales?

if "net_sales" in df.columns:
    mean_sales = df["net_sales"].mean()

    filtered_sales = df.loc[
        df["net_sales"] > mean_sales
    ]

    print("Question: Which orders have net_sales greater than the average?")
    print("Average net_sales:", mean_sales)
    print("Number of matching records:", len(filtered_sales))

    display(filtered_sales.head(10))
else:
    print("The net_sales column is not available.")
```

```python
# Question 2:
# Which orders generated a positive profit?

if "profit" in df.columns:
    profitable_orders = df.loc[
        df["profit"] > 0
    ]

    print("Question: Which orders generated a positive profit?")
    print("Number of profitable records:", len(profitable_orders))

    display(profitable_orders.head(10))
else:
    print("The profit column is not available.")
```

```python
# Question 3:
# Which orders belong to a specific region?

if "region" in df.columns:
    available_regions = df["region"].dropna().unique()

    if len(available_regions) > 0:
        chosen_region = available_regions[0]

        region_orders = df.loc[
            df["region"] == chosen_region
        ]

        print("Question: Which orders belong to the selected region?")
        print("Selected region:", chosen_region)
        print("Number of matching records:", len(region_orders))

        display(region_orders.head(10))
    else:
        print("No non-missing region values are available.")
else:
    print("The region column is not available.")
```

```python
# Question 4:
# Which records belong to repeat customers with positive profit?

if {"is_repeat_customer", "profit"}.issubset(df.columns):
    repeat_customer_values = df["is_repeat_customer"].dropna().unique()

    print("Available values for is_repeat_customer:")
    print(repeat_customer_values)

    # Handle Boolean, numeric, and text representations
    if pd.api.types.is_bool_dtype(df["is_repeat_customer"]):
        repeat_mask = df["is_repeat_customer"] == True
    elif pd.api.types.is_numeric_dtype(df["is_repeat_customer"]):
        repeat_mask = df["is_repeat_customer"] == 1
    else:
        repeat_mask = (
            df["is_repeat_customer"]
            .astype("string")
            .str.strip()
            .str.lower()
            .isin(["true", "yes", "1"])
        )

    repeat_profitable_orders = df.loc[
        repeat_mask & (df["profit"] > 0)
    ]

    print("\nQuestion: Which repeat-customer records have positive profit?")
    print("Number of matching records:", len(repeat_profitable_orders))

    display(repeat_profitable_orders.head(10))
else:
    print(
        "The required columns are not available. "
        "Available columns:", df.columns.tolist()
    )
```

```python
print("DATA QUALITY CHECK")
print("=" * 50)

print("Dataset shape:", df.shape)
print("Total missing values:", int(df.isna().sum().sum()))
print("Duplicate rows:", int(df.duplicated().sum()))

print("\nMissing values by column:")
display(df.isna().sum().sort_values(ascending=False).to_frame("Missing values"))

print("\nDuplicate records:")
display(df.loc[df.duplicated(keep=False)].head(10))

print("\nUnique values per column:")
display(df.nunique(dropna=False).to_frame("Unique values"))
```

```python
numeric_df = df.select_dtypes(include=np.number)

if numeric_df.shape[1] >= 2:
    correlation_matrix = numeric_df.corr()

    print("Correlation matrix:")
    display(correlation_matrix.round(3))

    plt.figure(figsize=(12, 8))

    image = plt.imshow(
        correlation_matrix,
        cmap="coolwarm",
        vmin=-1,
        vmax=1,
        aspect="auto"
    )

    plt.colorbar(image, label="Correlation coefficient")
    plt.xticks(
        range(len(correlation_matrix.columns)),
        correlation_matrix.columns,
        rotation=90
    )
    plt.yticks(
        range(len(correlation_matrix.index)),
        correlation_matrix.index
    )

    plt.title("Correlation Matrix of Numerical Variables")
    plt.tight_layout()
    plt.show()
else:
    print("At least two numerical columns are required for correlation analysis.")
```

```python
if {"net_sales", "profit"}.issubset(df.columns):
    sales_profit = df[["net_sales", "profit"]].dropna()

    correlation = sales_profit["net_sales"].corr(
        sales_profit["profit"]
    )

    print("Correlation between net_sales and profit:", round(correlation, 4))

    plt.figure(figsize=(8, 6))

    plt.scatter(
        sales_profit["net_sales"],
        sales_profit["profit"],
        alpha=0.4
    )

    plt.title("Relationship Between Net Sales and Profit")
    plt.xlabel("Net Sales")
    plt.ylabel("Profit")
    plt.tight_layout()
    plt.show()

    if correlation > 0:
        print("Interpretation: The variables have a positive linear association.")
    elif correlation < 0:
        print("Interpretation: The variables have a negative linear association.")
    else:
        print("Interpretation: No linear association is detected by the correlation coefficient.")
else:
    print("The net_sales and profit columns are required for this analysis.")
```

```python
print("FINAL DATASET SUMMARY")
print("=" * 50)

print("\nDataset dimensions:")
print(f"Rows: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]}")

print("\nNumerical variables:")
print(df.select_dtypes(include=np.number).columns.tolist())

print("\nCategorical variables:")
print(
    df.select_dtypes(
        include=["object", "category", "bool", "string"]
    ).columns.tolist()
)

print("\nData quality:")
print("Missing values:", int(df.isna().sum().sum()))
print("Duplicate rows:", int(df.duplicated().sum()))

if "region" in df.columns:
    print("\nRecords by region:")
    display(df["region"].value_counts(dropna=False))

if "is_repeat_customer" in df.columns:
    print("\nRepeat customer distribution:")
    display(df["is_repeat_customer"].value_counts(dropna=False))

if {"net_sales", "profit"}.issubset(df.columns):
    print("\nSales and profit summary:")
    display(df[["net_sales", "profit"]].describe())
```
